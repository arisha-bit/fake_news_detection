````python
import sys
sys.path.insert(0, '..')

from app.ml.image_classifier.train import train
import json
````

````python
# ── CONFIG ───────────────────────────────────────────────
DATA_DIR   = '../datasets/image_dataset'  # change to your dataset path
EPOCHS     = 10
BATCH_SIZE = 32
LR         = 1e-3
# ─────────────────────────────────────────────────────────
````

````python
metrics = train(
    data_dir=DATA_DIR,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    lr=LR,
)
print(json.dumps({k: v for k, v in metrics.items() if k != 'classification_report'}, indent=2))
print(metrics['classification_report'])
````

````python
# Plot training history
import matplotlib.pyplot as plt

history = metrics['history']
epochs  = [h['epoch'] for h in history]
accs    = [h['val_acc'] for h in history]

plt.figure(figsize=(8, 4))
plt.plot(epochs, accs, marker='o', color='indigo')
plt.title('Validation Accuracy per Epoch')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('results/image_classifier_training.png', dpi=120)
plt.show()
````

````python
# Confusion matrix
import seaborn as sns
import numpy as np

cm = np.array(metrics['confusion_matrix'])
labels = metrics['class_names']

plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=labels, yticklabels=labels,
            cmap='Blues', linewidths=0.5)
plt.title('Confusion Matrix')
plt.ylabel('True')
plt.xlabel('Predicted')
plt.tight_layout()
plt.savefig('results/image_classifier_cm.png', dpi=120)
plt.show()
````
