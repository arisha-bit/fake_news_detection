```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
fake_df = pd.read_csv("../datasets/Fake.csv")
real_df = pd.read_csv("../datasets/Real.csv")
```

```python
fake_df["label"] = 0
```

```python
real_df["label"] = 1
```

```python
df = pd.concat(
    [fake_df, real_df],
    ignore_index=True
)
```

```python
df.shape
```

```python
df["label"].value_counts()
```

```python
df["label"].value_counts().plot(
    kind="bar"
)

plt.show()
```

```python
df["content"] = (
    df["title"].fillna("")
    + " "
    + df["text"].fillna("")
)
```

```python
df.to_csv(
    "../datasets/processed/news.csv",
    index=False
)
```
